```python
import re
import json
import unicodedata
from num2words import num2words


def normalize_text(text):
    text = text.lower()

    # Convertir números a palabras en español
    text = re.sub(
        r'\d+',
        lambda match: num2words(int(match.group()), lang='es'),
        text
    )

    # Eliminar puntuación, conservando letras, tildes, ñ y espacios
    text = ''.join(
        char for char in text
        if char.isalpha() or char.isspace()
    )

    # Normalizar espacios
    text = ' '.join(text.split())

    return text
```

```python
class ViSpeRDataset(Dataset):
    def __init__(self, dataframe):
        self.dataframe = dataframe.reset_index(drop=True)

    def __len__(self):
        return len(self.dataframe)

    def __getitem__(self, idx):
        row = self.dataframe.iloc[idx]

        cap = cv2.VideoCapture(str(row["path"]))

        if not cap.isOpened():
            raise RuntimeError(f"No se pudo abrir: {row['path']}")

        frames = []

        while True:
            ret, frame = cap.read()

            if not ret:
                break

            frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
            frames.append(frame)

        cap.release()

        if not frames:
            raise RuntimeError(f"El clip no contiene frames: {row['path']}")

        video = np.stack(frames).astype(np.float32) / 255.0
        video = torch.from_numpy(video).permute(0, 3, 1, 2)

        target = torch.tensor(
            [char_to_idx[char] for char in str(row["label"])],
            dtype=torch.long
        )

        return {
            "video": video,
            "target": target,
            "input_length": video.shape[0],
            "target_length": target.numel(),
            "text": str(row["label"]),
            "path": str(row["path"])
        }
```

# DEFINIR EL MODELO

```python
class MaskedBatchNorm3d(nn.Module):
    def __init__(self, num_features, eps=1e-5, momentum=0.1):
        super().__init__()
        self.last_batch_mean = None
        self.last_batch_var = None

        self.eps = eps
        self.momentum = momentum

        self.weight = nn.Parameter(torch.ones(num_features))
        self.bias = nn.Parameter(torch.zeros(num_features))

        self.register_buffer(
            "running_mean", torch.zeros(num_features)
        )
        self.register_buffer(
            "running_var", torch.ones(num_features)
        )

    def forward(self, x, lengths):
        # x: (B, C, T, H, W)
        B, C, T, H, W = x.shape

        # Máscara temporal: (B, 1, T, 1, 1)
        valid = (
            torch.arange(T, device=x.device)[None, :]
            < lengths.to(x.device)[:, None]
        )

        mask = valid[:, None, :, None, None].to(x.dtype)

        if self.training:
            count = mask.sum() * H * W

            mean = (x * mask).sum(dim=(0, 2, 3, 4)) / count

            var = (
                ((x - mean[None, :, None, None, None]) ** 2) * mask
            ).sum(dim=(0, 2, 3, 4)) / count

            with torch.no_grad():
                n = count.item()
                if n > 1:
                    unbiased_var = var.detach() * (n / (n - 1))
                else:
                    unbiased_var = var.detach()

                self.running_mean.lerp_(mean.detach(), self.momentum)
                self.running_var.lerp_(unbiased_var, self.momentum)

        else:
            mean = self.running_mean
            var = self.running_var

        x = (
            (x - mean[None, :, None, None, None])
            / torch.sqrt(var[None, :, None, None, None] + self.eps)
        )

        x = (
            x * self.weight[None, :, None, None, None]
            + self.bias[None, :, None, None, None]
        )

        self.last_batch_mean = mean.detach()
        self.last_batch_var = var.detach()

        return x * mask
```

```python
class STCNN(nn.Module):
    def __init__(self, in_channels=3):
        super().__init__()

        # BLOCK 1
        self.bn1 = MaskedBatchNorm3d(in_channels)

        self.conv1 = nn.Conv3d(
            in_channels, 32,
            kernel_size=(3, 5, 5),
            stride=(1, 2, 2),
            padding=(1, 2, 2)
        )

        self.bn2 = MaskedBatchNorm3d(32)
        self.relu1 = nn.ReLU()

        self.pool1 = nn.MaxPool3d(
            kernel_size=(1, 2, 2),
            stride=(1, 2, 2)
        )

        self.dropout1 = nn.Dropout3d(p=0.1)

        # BLOCK 2
        self.conv2 = nn.Conv3d(
            32, 64,
            kernel_size=(3, 5, 5),
            stride=(1, 1, 1),
            padding=(1, 2, 2)
        )

        self.bn3 = MaskedBatchNorm3d(64)
        self.relu2 = nn.ReLU()

        self.pool2 = nn.MaxPool3d(
            kernel_size=(1, 2, 2),
            stride=(1, 2, 2)
        )

        self.dropout2 = nn.Dropout3d(p=0.1)

        # BLOCK 3
        self.conv3 = nn.Conv3d(
            64, 96,
            kernel_size=(3, 3, 3),
            stride=(1, 1, 1),
            padding=(1, 1, 1)
        )

        self.bn4 = MaskedBatchNorm3d(96)
        self.relu3 = nn.ReLU()

        self.pool3 = nn.MaxPool3d(
            kernel_size=(1, 2, 2),
            stride=(1, 2, 2)
        )

    def apply_temporal_mask(self, x, lengths):
        temporal_size = x.size(2)

        valid_positions = (
            torch.arange(temporal_size, device=x.device)[None, :]
            < lengths.to(x.device)[:, None]
        )

        mask = valid_positions[:, None, :, None, None]

        return x * mask

    def forward(self, x, lengths):

        # BLOCK 1
        x = self.bn1(x, lengths)
        x = self.conv1(x)
        x = self.bn2(x, lengths)
        x = self.relu1(x)
        x = self.pool1(x)
        x = self.dropout1(x)
        x = self.apply_temporal_mask(x, lengths)

        # BLOCK 2
        x = self.conv2(x)
        x = self.bn3(x, lengths)
        x = self.relu2(x)
        x = self.pool2(x)
        x = self.dropout2(x)
        x = self.apply_temporal_mask(x, lengths)

        # BLOCK 3
        x = self.conv3(x)
        x = self.bn4(x, lengths)
        x = self.relu3(x)
        x = self.pool3(x)
        x = self.apply_temporal_mask(x, lengths)

        return x
```

```python
class TemporalEncoder(nn.Module):
    def __init__(
        self,
        input_size=3456,
        hidden_size=256,
        num_layers=2,
        rnn_type="gru"
    ):
        super().__init__()

        if rnn_type.lower() == "gru":
            rnn_class = nn.GRU
        elif rnn_type.lower() == "lstm":
            rnn_class = nn.LSTM
        else:
            raise ValueError("rnn_type debe ser 'gru' o 'lstm'")

        self.rnn = rnn_class(
            input_size=input_size,
            hidden_size=hidden_size,
            num_layers=num_layers,
            bidirectional=True,
            batch_first=False
        )

    def forward(self, x, lengths):
        packed = pack_padded_sequence(
            x,
            lengths.cpu(),
            batch_first=False,
            enforce_sorted=False
        )

        packed_output, _ = self.rnn(packed)

        output, _ = pad_packed_sequence(
            packed_output,
            batch_first=False,
            total_length=x.size(0)
        )

        return output
```

```python
class CharacterClassifier(nn.Module):
    def __init__(self, input_size=512, vocab_size=VOCAB_SIZE):
        super().__init__()
        self.linear = nn.Linear(input_size, vocab_size)

    def forward(self, x):
        return self.linear(x)
```

#cosillas:
la entrada es lo que devuelve el dataLoader -> $(B,T,C,H,W)$
La salida queda en (T,B,V), que es el formato que quiere CTCLoss

```python
class LipReadingModel(nn.Module): 
    def __init__(
        self,
        in_channels=3,
        vocab_size=VOCAB_SIZE,
        hidden_size=256,
        rnn_type="gru"
    ):
        super().__init__()

        self.visual_encoder = STCNN(in_channels=in_channels)

        self.temporal_encoder = TemporalEncoder(
            input_size=96 * 6 * 6,
            hidden_size=hidden_size,
            num_layers=2,
            rnn_type=rnn_type
        )

        self.classifier = CharacterClassifier(
            input_size=hidden_size * 2,
            vocab_size=vocab_size
        )

    def forward(self, x, input_lengths):
        # B,T,C,H,W -> B,C,T,H,W
        x = x.permute(0, 2, 1, 3, 4)

        # Extracción de características espacio-temporales
        x = self.visual_encoder(x, input_lengths)

        # B,C,T,H,W -> T,B,C*H*W
        x = x.permute(2, 0, 1, 3, 4)
        x = x.reshape(x.shape[0], x.shape[1], -1)

        # Modelado temporal respetando las longitudes reales
        x = self.temporal_encoder(x, input_lengths)

        # Clasificación por timestep
        x = self.classifier(x)

        return torch.log_softmax(x, dim=-1)
```

```python
def collate_fn(batch):
    videos = [item["video"] for item in batch]
    targets = [item["target"] for item in batch]

    input_lengths = torch.tensor(
        [video.shape[0] for video in videos],
        dtype=torch.long
    )

    target_lengths = torch.tensor(
        [target.numel() for target in targets],
        dtype=torch.long
    )

    max_frames = input_lengths.max().item()

    padded_videos = torch.zeros(
        len(batch),
        max_frames,
        *videos[0].shape[1:],
        dtype=torch.float32
    )

    for i, video in enumerate(videos):
        padded_videos[i, :video.shape[0]] = video

    concatenated_targets = torch.cat(targets)

    return {
        "videos": padded_videos,
        "targets": concatenated_targets,
        "input_lengths": input_lengths,
        "target_lengths": target_lengths,
        "texts": [item["text"] for item in batch],
        "paths": [item["path"] for item in batch]
    }
```

```python
# ---------- helpers para ver qué está prediciendo ----------
def tok_to_symbol(tok):
    """blank -> '·' | espacio -> '␣' | resto -> el caracter."""
    if tok == BLANK_IDX:
        return "·"
    ch = idx_to_char[tok]
    return "␣" if ch == " " else ch


def raw_frames(tokens):
    """Argmax por frame SIN colapsar: lo que la red dice en cada frame."""
    return "".join(tok_to_symbol(t) for t in tokens)


def ctc_collapse(tokens):
    """Greedy CTC: colapsa repetidos y elimina blanks."""
    out, prev = [], None
    for t in tokens:
        if t != prev and t != BLANK_IDX:
            out.append(idx_to_char[t])
        prev = t
    return "".join(out)


def min_frames_needed(target_tokens):
    """Frames mínimos para alinear el target: 1 por caracter
    + 1 blank obligatorio entre cada par de caracteres iguales seguidos."""
    repeats = sum(1 for a, b in zip(target_tokens[:-1], target_tokens[1:]) if a == b)
    return len(target_tokens) + repeats


def top_tokens(tokens, k=3):
    total = len(tokens)
    return " ".join(
        f"{tok_to_symbol(t)}:{100 * n / total:.0f}%"
        for t, n in Counter(tokens).most_common(k)
    )
```

```python
def greedy_decode(log_probs, input_length):
    """
    Convierte las predicciones por timestep en una secuencia
    de caracteres usando greedy decoding + CTC collapse.
    """

    predictions = log_probs[:input_length].argmax(dim=-1).tolist()

    decoded = []
    previous = None

    for token in predictions:

        if token != previous and token != BLANK_IDX:
            decoded.append(idx_to_char[token])

        previous = token

    return "".join(decoded)
```

```python
def edit_distance(reference, hypothesis):
    """
    Distancia de Levenshtein:
    sustituciones + inserciones + eliminaciones.
    """

    rows = len(reference) + 1
    cols = len(hypothesis) + 1

    dp = [[0] * cols for _ in range(rows)]

    for i in range(rows):
        dp[i][0] = i

    for j in range(cols):
        dp[0][j] = j

    for i in range(1, rows):
        for j in range(1, cols):

            if reference[i - 1] == hypothesis[j - 1]:
                cost = 0
            else:
                cost = 1

            dp[i][j] = min(
                dp[i - 1][j] + 1,       # Eliminación
                dp[i][j - 1] + 1,       # Inserción
                dp[i - 1][j - 1] + cost  # Sustitución
            )

    return dp[-1][-1]
```

# TRAIN loop

```python
history = []         # loss promedio de train por época
val_history = []     # loss promedio de validation por época
iter_history = []    # loss de cada iteración

best_val_loss = float("inf")
best_val_cer = float("inf")

n_iters = len(train_loader)

epochs = start_epoch + EPOCHS

for epoch in range(start_epoch, epochs):

    # =========================================================
    # TRAIN
    # =========================================================
    model.train()
    total_loss = 0.0
    valid_batches = 0
    seen_samples = 0
    empty_predictions = 0
    impossible_samples = 0
    skipped_nan = 0

    for batch_idx, batch in enumerate(train_loader):

        videos = batch["videos"].to(DEVICE)
        targets = batch["targets"].to(DEVICE)
        input_lengths = batch["input_lengths"]
        target_lengths = batch["target_lengths"]

        optimizer.zero_grad()

        log_probs = model(videos, input_lengths)      # (T, B, V)

        loss = ctc_loss(
            log_probs,
            targets,
            input_lengths,
            target_lengths
        )

        if not torch.isfinite(loss):
            skipped_nan += 1
            print(f"[época {epoch + 1} | iter {batch_idx + 1}] loss no finita, se saltea")
            continue

        loss.backward()

        grad_norm = torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            max_norm=5.0
        )

        optimizer.step()

        total_loss += loss.item()
        valid_batches += 1
        iter_history.append(loss.item())

        # ----- contadores + inspección -----
        preds = log_probs.detach().argmax(dim=-1).cpu()
        target_offset = 0

        for i in range(len(input_lengths)):
            T_i = input_lengths[i].item()
            L_i = target_lengths[i].item()

            target = batch["targets"][target_offset : target_offset + L_i].tolist()
            target_offset += L_i

            need = min_frames_needed(target)
            impossible = need > T_i
            if impossible:
                impossible_samples += 1

            frame_tokens = preds[:T_i, i].tolist()
            prediction = ctc_collapse(frame_tokens)

            if prediction == "":
                empty_predictions += 1

            seen_samples += 1

            # Inspección solo del primer clip del batch y cada PRINT_CADA
            if i == 0 and (batch_idx + 1) % PRINT_CADA == 0:
                reference = "".join(idx_to_char[t] for t in target)
                blank_pct = 100 * frame_tokens.count(BLANK_IDX) / max(T_i, 1)

                print(
                    f"\n[época {epoch + 1}/{epochs} | iter {batch_idx + 1}/{n_iters}] "
                    f"loss={loss.item():.4f} | grad_norm={float(grad_norm):.2f}"
                )
                print(
                    f"  {Path(batch['paths'][i]).name} | "
                    f"frames={T_i} | caracteres={L_i} (mín. {need})"
                )
                if impossible:
                    print("  ⚠ Clip incompatible con CTC")
                print(f"  REAL: {reference!r}")
                print(f"  PRED: {prediction!r}")
                print(f"  blank={blank_pct:.0f}%")

    # =========================================================
    # VALIDATION
    # =========================================================
    model.eval()

    val_total_loss = 0.0
    val_valid_batches = 0
    val_char_errors = 0
    val_total_chars = 0
    val_empty_predictions = 0
    val_impossible = 0
    val_skipped_nan = 0

    with torch.no_grad():
        for batch in val_loader:

            videos = batch["videos"].to(DEVICE)
            targets = batch["targets"].to(DEVICE)
            input_lengths = batch["input_lengths"]
            target_lengths = batch["target_lengths"]

            log_probs = model(videos, input_lengths)

            loss = ctc_loss(
                log_probs,
                targets,
                input_lengths,
                target_lengths
            )

            if not torch.isfinite(loss):
                val_skipped_nan += 1
                continue

            val_total_loss += loss.item()
            val_valid_batches += 1

            # ----- CER -----
            preds = log_probs.detach().argmax(dim=-1).cpu()
            target_offset = 0

            for i in range(len(input_lengths)):
                T_i = input_lengths[i].item()
                L_i = target_lengths[i].item()

                target = targets[target_offset : target_offset + L_i].tolist()
                target_offset += L_i

                if min_frames_needed(target) > T_i:
                    val_impossible += 1
                    continue

                prediction = greedy_decode(log_probs[:, i, :], T_i)
                reference = "".join(idx_to_char[token] for token in target)

                reference_norm = normalize_text(reference)
                prediction_norm = normalize_text(prediction)

                val_char_errors += edit_distance(reference_norm, prediction_norm)
                val_total_chars += len(reference_norm)
                val_empty_predictions += int(prediction_norm == "")

    val_loss = val_total_loss / max(val_valid_batches, 1)
    val_CER = (
        val_char_errors / val_total_chars if val_total_chars > 0 else float("nan")
    )

    # =========================================================
    # CHECKPOINT
    # =========================================================
    improved = False

    if val_loss < best_val_loss:
        best_val_loss = val_loss
        improved = True

    if not math.isnan(val_CER) and val_CER < best_val_cer:
        best_val_cer = val_CER
        improved = True

    if improved:
        checkpoint_path = CHECKPOINT_DIR / "best_model.pth"
        torch.save({
            "epoch": epoch + 1,
            "model_state_dict": model.state_dict(),
            "optimizer_state_dict": optimizer.state_dict(),
            "best_val_loss": best_val_loss,
            "best_val_cer": best_val_cer,
            "val_loss": val_loss,
            "val_cer": val_CER,
            "vocab_size": VOCAB_SIZE,
            "blank_idx": BLANK_IDX,
        }, checkpoint_path)
        print(f"  ✓ Checkpoint guardado | val_loss={best_val_loss:.4f} | val_CER={best_val_cer:.4f}")

    # =========================================================
    # LOGS DE ÉPOCA
    # =========================================================
    epoch_loss = total_loss / max(valid_batches, 1)
    history.append(epoch_loss)
    val_history.append(val_loss)

    print("\n" + "=" * 70)
    print(
        f"Época {epoch + 1}/{epochs} | "
        f"Train Loss: {epoch_loss:.4f} | "
        f"Val Loss: {val_loss:.4f} | "
        f"Batches: {valid_batches}/{n_iters}"
    )
    print(f"  Predicciones vacías (train): {empty_predictions}/{seen_samples}")
    print(f"  Clips imposibles CTC (train): {impossible_samples}/{seen_samples}")
    if skipped_nan > 0:
        print(f"  Batches salteados por NaN/Inf (train): {skipped_nan}")
    print(f"  Validation CER: {val_CER:.4f} ({val_CER * 100:.2f}%)")
    print(f"  Predicciones vacías (val): {val_empty_predictions}")
    print(f"  Clips imposibles CTC (val): {val_impossible}")
    if val_skipped_nan > 0:
        print(f"  Batches salteados por NaN/Inf (val): {val_skipped_nan}")
    print("=" * 70)
```

## eval

```python
model.eval()

total_loss = 0.0
valid_batches = 0

total_char_errors = 0
total_chars = 0

total_word_errors = 0
total_words = 0

val_char_errors = 0
val_total_chars = 0
val_empty_predictions = 0

results = []

with torch.no_grad():

    for batch in test_loader:

        videos = batch["videos"].to(DEVICE)
        targets = batch["targets"].to(DEVICE)

        input_lengths = batch["input_lengths"]
        target_lengths = batch["target_lengths"]

        # Forward
        log_probs = model(videos, input_lengths)

        # CTC Loss
        loss = ctc_loss(
            log_probs,
            targets,
            input_lengths,
            target_lengths
        )

        if torch.isfinite(loss):
            total_loss += loss.item()
            valid_batches += 1

        # Decodificar cada video del batch
        target_offset = 0

        for i in range(len(input_lengths)):

            # Predicción
            prediction = greedy_decode(
                log_probs[:, i, :],
                input_lengths[i].item()
            )

            # Target correspondiente al video
            target_length = target_lengths[i].item()

            target = targets[
                target_offset:target_offset + target_length
            ].tolist()

            reference = "".join(
                idx_to_char[token]
                for token in target
            )

            target_offset += target_length

            # Normalización
            reference_norm = normalize_text(reference)
            prediction_norm = normalize_text(prediction)

            # CER
            char_errors = edit_distance(
                reference_norm,
                prediction_norm
            )

            total_char_errors += char_errors
            total_chars += len(reference_norm)

            # WER
            reference_words = reference_norm.split()
            prediction_words = prediction_norm.split()

            word_errors = edit_distance(
                reference_words,
                prediction_words
            )

            total_word_errors += word_errors
            total_words += len(reference_words)

            results.append({
                "reference": reference,
                "prediction": prediction,
                "char_errors": char_errors,
                "word_errors": word_errors
            })
```

```python
test_loss = (
    total_loss / valid_batches
    if valid_batches > 0 else float("nan")
)

CER = (
    total_char_errors / total_chars
    if total_chars > 0 else float("nan")
)

WER = (
    total_word_errors / total_words
    if total_words > 0 else float("nan")
)
```

CTC LOSS Y GREEDY DECODING NO predicen exactamente lo mismo  
Ctc optimiza la probabilidad total de todas las alineaciones posibles con la transcripcion correcta 

greedy decoding selecciona el caracter mas probable de cada frame, elimina blanks y colapsa caracteres repetidos
entonces una mejora en CTC No garantiza necesariamente una inmediata en CER
